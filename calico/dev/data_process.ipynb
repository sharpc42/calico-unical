```python
num = 120
pairs = int(0.5 * (num**2 - num))
output_vals = np.zeros((num,))
for i in range(120):
    np.random.seed(i * 500)
    m_r = np.random.normal(loc=0, scale=5, size=pairs)
    m_i = np.random.normal(loc=0, scale=5, size=pairs)
    e_r = np.random.normal(loc=0, scale=0.2, size=pairs)
    e_i = np.random.normal(loc=0, scale=0.2, size=pairs)
    n_r = np.random.normal(loc=0, scale=0.1, size=pairs)
    n_i = np.random.normal(loc=0, scale=0.1, size=pairs)
    m = m_r + 1.0j * m_i
    e = e_r + 1.0j * e_i
    n = n_r + 1.0j * n_i
    v = m + e + n
    first_term = np.mean((np.conj(m) * v).real)
    second_term = np.mean(np.abs(m) ** 2)
    output_vals[i] = first_term - second_term
print(np.mean(np.asarray(output_vals)))
```
